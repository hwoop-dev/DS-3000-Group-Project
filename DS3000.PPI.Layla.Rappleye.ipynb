2. The data Source I am using data I found from the USGS Water Data API. Using this I will collect observations from their monitoring stations. The data can be pulled directly and fulfills requirements

```python
import requests
```

```python
url = "https://api.waterdata.usgs.gov/ogcapi/v1/collections/daily/items"
```

```python
# pulling streamflow data from a working example site to check working API
streamflow_params = {"monitoring_location_id": "USGS-01435000",
    "parameter_code": "00060",
    "statistic_id": "00003",
    "datetime": "2025-01-01/2025-12-31",
    "limit": 500}
streamflow_response = requests.get(url, params = streamflow_params)
print("Status code:", streamflow_response.status_code)

streamflow_data = streamflow_response.json()
print("Number of streamflow observations:", len(streamflow_data["features"]))

# displaying first five observations to check the fields returned
for feature in streamflow_data["features"][:5]:
    print(feature["properties"])
```
